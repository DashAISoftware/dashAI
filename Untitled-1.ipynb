```python
def generate_chunks(texts, size, overlap)-> Dict[int, List[str]]:
    """
    Generate chunks of text from the provided texts.
    Args:
        texts (List[str]): List of text documents to chunk.
        size (int): Size of each chunk.
        overlap (int): Overlap size between chunks.
    Returns:
        Dict[int, List[str]]: Dictionary with document index as key and list of chunks as value.
    """
    
    documents_chunks = {}

    for i, text in enumerate(texts):
        chunks = []
        for j in range(0, len(text), size - overlap):
            if j + size > len(text):
                chunk = text[j:]
            else:
                chunk = text[j:j + size]
            chunks.append(chunk)
        documents_chunks[i] = chunks
    return documents_chunks
```
